```python
#Rosalind 5
word_dict = {}
with open("practice_5.txt", "r") as file:
  for line in file:
    for word in line.split():
      if word in word_dict:
        word_dict[word] = word_dict[word] + 1
      else:
        word_dict[word] = 1
print(word_dict)
```

```python
word_dict={}
with open("practice_5.txt", "r") as file:
  words=[]
  for line in file:
    words.extend(line.split())
  for word in words:
    word_dict[word] = words.count(word)
print(word_dict)
```
